```python
#checking for wrong values:
from classification.validation import invalid_category, invalid_number

data_issues_summary = {
    "Age": invalid_number(raw_df["Age"], 1, 120, integer=True).sum(),
    "Gender": invalid_category(raw_df["Gender"], ["Male", "Female", "Non-binary"]).sum(),
    "Year_of_Study": invalid_number(raw_df["Year_of_Study"], 1, 5, integer=True).sum(),
    "Sleep_Hours": invalid_number(raw_df["Sleep_Hours"], 0, 24).sum(),
    "Study_Hours_Per_Week": invalid_number(raw_df["Study_Hours_Per_Week"], 0, 168).sum(),
    "GPA": invalid_number(raw_df["GPA"], 0, 4).sum(),
    "Social_Media_Hours": invalid_number(raw_df["Social_Media_Hours"], 0, 24).sum(),
    "Screen_Time_Hours": invalid_number(raw_df["Screen_Time_Hours"], 0, 24).sum(),
    "Physical_Activity": invalid_category(
        raw_df["Physical_Activity"], ["Low", "Moderate", "High"]
    ).sum(),
    "Financial_Stress": invalid_number(raw_df["Financial_Stress"], 1, 10).sum(),
    "Family_Support": invalid_number(raw_df["Family_Support"], 1, 10).sum(),
    "Attendance_Rate": invalid_number(raw_df["Attendance_Rate"], 0, 100).sum(),
    "Part_Time_Job": invalid_category(raw_df["Part_Time_Job"], ["Yes", "No"]).sum(),
    "Diet_Quality": invalid_category(raw_df["Diet_Quality"], ["Poor", "Average", "Good"]).sum(),
    "Previous_Mental_Health_History": invalid_category(
        raw_df["Previous_Mental_Health_History"], ["Yes", "No"]
    ).sum(),
    "Mental_Health_Risk": invalid_category(
        raw_df["Mental_Health_Risk"], ["Low", "Medium", "High"]
    ).sum(),
}

display( pd.DataFrame([data_issues_summary], index = ["duplicated"]).T)
```
